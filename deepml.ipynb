```python
class ImageNeuralNetwork(nn.Module):
    def __init__(self, width, height):
        super().__init__() # <-- IMPORTANT
        self.flatten = nn.Flatten()
        self.model = nn.Sequential(
        nn.Linear(width * height, 512), nn.ReLU(),
        nn.Linear(512, 128), nn.ReLU(),
        nn.Linear(128, 1)
        )
    def forward(self, x):
        x = self.flatten(x) # Flatten the input data
        y = self.model(x) # Call the data on the flatten input `x`
        return y
```
